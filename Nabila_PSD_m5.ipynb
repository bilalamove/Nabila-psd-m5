```python
import pandas as pd
import numpy as np
```

```python
# Membaca dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# Menampilkan 5 baris pertama
df.head()
```

```python
# Menampilkan informasi struktur dataset
df.info()
```

```python
# Menampilkan jumlah baris dan kolom
print("Jumlah baris dan kolom:", df.shape)

# Menampilkan nama seluruh kolom
print("\nNama kolom:")
print(df.columns.tolist())
```

```python
# Menghitung jumlah missing value setiap kolom
missing = df.isnull().sum()

print("Jumlah Missing Value:")
print(missing)

# Menampilkan kolom yang memiliki missing value
print("\nKolom yang memiliki Missing Value:")
print(missing[missing > 0])
```

```python
# Mengisi kode kasir yang kosong
df['kode_kasir'] = df['kode_kasir'].fillna('TIDAK DIKETAHUI')

# Menghapus baris yang memiliki data penting kosong
kolom_penting = [
    'tanggal',
    'jumlah',
    'harga_satuan',
    'diskon_persen',
    'status'
]

df = df.dropna(subset=kolom_penting)

# Menghapus data duplikat
df = df.drop_duplicates()

print("Missing Value setelah penanganan:")
print(df.isnull().sum())
```

```python
# Mengubah kolom tanggal menjadi datetime
df['tanggal'] = pd.to_datetime(
    df['tanggal'],
    errors='coerce'
)

# Mengubah kolom numerik
kolom_numerik = [
    'jumlah',
    'harga_satuan',
    'diskon_persen'
]

for kolom in kolom_numerik:
    df[kolom] = pd.to_numeric(
        df[kolom],
        errors='coerce'
    )

# Menghapus data yang gagal dikonversi
df = df.dropna(
    subset=['tanggal', 'jumlah', 'harga_satuan', 'diskon_persen']
)

# Menampilkan tipe data terbaru
df.info()
```

```python
# Menyeragamkan penulisan status
df['status'] = df['status'].astype(str).str.strip().str.upper()

# Melihat jumlah masing-masing status
print(df['status'].value_counts())

# Melihat status yang tersedia
print("\nStatus transaksi:")
print(df['status'].unique())
```

```python
# Menampilkan informasi diskon
print("Diskon terendah:", df['diskon_persen'].min())
print("Diskon tertinggi:", df['diskon_persen'].max())

# Memeriksa diskon yang tidak valid
print("\nDiskon di luar rentang 0-100:")
print(df[
    (df['diskon_persen'] < 0) |
    (df['diskon_persen'] > 100)
])
```

```python
# Menghapus transaksi dengan diskon tidak valid
df = df[
    (df['diskon_persen'] >= 0) &
    (df['diskon_persen'] <= 100)
].copy()

# Menghitung nilai transaksi sebelum diskon
df['total_harga'] = (
    df['jumlah'] * df['harga_satuan']
)

# Menghitung nilai setelah diskon
df['nilai_setelah_diskon'] = (
    df['total_harga'] *
    (1 - df['diskon_persen'] / 100)
)

# Membulatkan hasil menjadi dua angka desimal
df['nilai_setelah_diskon'] = (
    df['nilai_setelah_diskon'].round(2)
)

# Menampilkan hasil perhitungan
df[[
    'jumlah',
    'harga_satuan',
    'diskon_persen',
    'total_harga',
    'nilai_setelah_diskon'
]].head(10)
```

```python
# Menghitung pendapatan bersih berdasarkan status
df['pendapatan_bersih'] = np.select(
    [
        df['status'] == 'SELESAI',
        df['status'] == 'RETUR',
        df['status'] == 'BATAL'
    ],
    [
        df['nilai_setelah_diskon'],
        -df['nilai_setelah_diskon'],
        0
    ],
    default=np.nan
)

# Menghapus status yang tidak dikenal
df = df.dropna(subset=['pendapatan_bersih'])

# Menampilkan hasil
df[[
    'status',
    'total_harga',
    'diskon_persen',
    'nilai_setelah_diskon',
    'pendapatan_bersih'
]].head(15)
```

```python
# Mengambil tahun dari kolom tanggal
df['tahun'] = df['tanggal'].dt.year

# Melihat jumlah transaksi berdasarkan tahun
print(df['tahun'].value_counts().sort_index())
```

```python
# Memilih data tahun 2024 dan 2025
df_tahunan = df[
    df['tahun'].isin([2024, 2025])
]

# Menghitung total pendapatan bersih per tahun
pendapatan_tahunan = (
    df_tahunan
    .groupby('tahun')['pendapatan_bersih']
    .sum()
    .reindex([2024, 2025], fill_value=0)
    .reset_index()
)

# Mengubah nama kolom
pendapatan_tahunan.columns = [
    'Tahun',
    'Total Pendapatan Bersih'
]

# Menampilkan hasil
print("TOTAL PENDAPATAN PER TAHUN")
display(pendapatan_tahunan)
```

```python
# Mengambil nilai pendapatan masing-masing tahun
pendapatan_2024 = pendapatan_tahunan.loc[
    pendapatan_tahunan['Tahun'] == 2024,
    'Total Pendapatan Bersih'
].iloc[0]

pendapatan_2025 = pendapatan_tahunan.loc[
    pendapatan_tahunan['Tahun'] == 2025,
    'Total Pendapatan Bersih'
].iloc[0]

# Menghitung selisih pendapatan
selisih = pendapatan_2025 - pendapatan_2024

print("Pendapatan 2024 : Rp{:,.2f}".format(pendapatan_2024))
print("Pendapatan 2025 : Rp{:,.2f}".format(pendapatan_2025))
print("Selisih         : Rp{:,.2f}".format(selisih))

# Menghitung persentase perubahan
if pendapatan_2024 != 0:
    persentase = (selisih / pendapatan_2024) * 100
    print("Perubahan       : {:.2f}%".format(persentase))
else:
    print("Perubahan       : Tidak dapat dihitung")
```

```python
# Menyimpan rekap pendapatan tahunan
pendapatan_tahunan.to_csv(
    'rekap_pendapatan_tahunan.csv',
    index=False
)

print("Rekap pendapatan tahunan berhasil disimpan!")
```
